```python
"""Cached confidence and tiled inference experiments, pinned to Ultralytics 8.4.165."""
from pathlib import Path
import json
import time

import cv2
import numpy as np
import torch
from torchvision.ops import batched_nms
from ultralytics import YOLO
from ultralytics.engine.validator import BaseValidator
from ultralytics.utils.metrics import DetMetrics, box_iou
from ultralytics.utils.ops import xywh2xyxy


def tile_starts(length, size, overlap=.25):
    if length <= size:
        return [0]
    stride = max(1, round(size * (1 - overlap)))
    return sorted(set(list(range(0, length - size + 1, stride)) + [length - size]))


def merge_boxes(predictions, iou=.7, max_det=300):
    if not len(predictions):
        return predictions.reshape(0, 6)
    data = torch.as_tensor(predictions, dtype=torch.float32)
    keep = batched_nms(data[:, :4], data[:, 4], data[:, 5], iou)[:max_det]
    return data[keep].numpy()


def ground_truth(label, width, height):
    text = Path(label).read_text(encoding='utf-8').strip()
    labels = np.array([list(map(float, line.split())) for line in text.splitlines()], dtype=np.float32).reshape(-1, 5)
    boxes = xywh2xyxy(labels[:, 1:5] * np.array([width, height, width, height]))
    return labels[:, 0], boxes.astype(np.float32)


def match(predictions, classes, boxes):
    validator = object.__new__(BaseValidator)
    validator.iouv = torch.linspace(.5, .95, 10)
    pred = torch.as_tensor(predictions, dtype=torch.float32)
    targets = torch.as_tensor(boxes, dtype=torch.float32)
    return validator.match_predictions(pred[:, 5], torch.as_tensor(classes), box_iou(targets, pred[:, :4])).numpy()


def summarize_cache(paths, names, merge_iou=None, thresholds=(.01, .03, .05, .1, .15, .2, .25, .3, .4, .5)):
    """AP uses all predictions >=.001. Fixed confidence P/R/F1 are separately re-matched at IoU .5."""
    metrics = DetMetrics(names=dict(enumerate(names)))
    counts = {threshold: np.zeros((len(names), 3), dtype=np.int64) for threshold in thresholds}
    latencies = []
    for path in paths:
        with np.load(path, allow_pickle=False) as cache:
            pred, classes, boxes = cache['pred'], cache['classes'], cache['boxes']
            elapsed = float(cache['seconds'])
        before_merge = time.perf_counter()
        if merge_iou is not None:
            pred = merge_boxes(pred, merge_iou)
        latencies.append(elapsed + time.perf_counter() - before_merge)
        correct = match(pred, classes, boxes)
        metrics.update_stats(dict(tp=correct, conf=pred[:, 4], pred_cls=pred[:, 5], target_cls=classes,
                                  target_img=np.unique(classes), im_name=path.stem))
        target_count = np.bincount(classes.astype(int), minlength=len(names))
        for threshold in thresholds:
            selected = pred[pred[:, 4] >= threshold]
            tp_mask = match(selected, classes, boxes)[:, 0]
            predicted = np.bincount(selected[:, 5].astype(int), minlength=len(names))
            tp = np.bincount(selected[tp_mask, 5].astype(int), minlength=len(names))
            counts[threshold] += np.stack([tp, predicted - tp, target_count - tp], axis=1)
    metrics.process(plot=False)
    operating_points = []
    for threshold, values in counts.items():
        tp, fp, fn = values.T
        p = np.divide(tp, tp + fp, out=np.zeros(len(names)), where=(tp + fp) > 0)
        r = np.divide(tp, tp + fn, out=np.zeros(len(names)), where=(tp + fn) > 0)
        f1 = np.divide(2*p*r, p+r, out=np.zeros(len(names)), where=(p+r) > 0)
        totals = values.sum(axis=0)
        micro_p = float(totals[0] / max(totals[0] + totals[1], 1))
        micro_r = float(totals[0] / max(totals[0] + totals[2], 1))
        operating_points.append(dict(confidence=threshold, match_iou=.5, precision_macro=float(p.mean()),
                                     recall_macro=float(r.mean()), f1_macro=float(f1.mean()),
                                     precision_micro=micro_p, recall_micro=micro_r,
                                     f1_micro=2*micro_p*micro_r/max(micro_p+micro_r, 1e-12),
                                     per_class=[dict(name=n, precision=float(a), recall=float(b), f1=float(c),
                                                     TP=int(v[0]), FP=int(v[1]), FN=int(v[2]))
                                                for n, a, b, c, v in zip(names, p, r, f1, values)]))
    return dict(mAP50=float(metrics.box.map50), mAP50_95=float(metrics.box.map),
                validator_precision_macro=float(metrics.box.mp), validator_recall_macro=float(metrics.box.mr),
                validator_f1_macro=float(metrics.box.f1.mean()),
                per_class_ap=json.loads(json.dumps(metrics.summary(), default=lambda v: v.item())),
                confidence_sweep=operating_points, mean_seconds=float(np.mean(latencies)),
                p95_seconds=float(np.percentile(latencies, 95)), images=len(paths),
                merge_iou=merge_iou, ap_confidence_floor=.001, max_det=300,
                note='Single-image inference evaluator; compare tile/full within this protocol, separately from native batch8 validation.')


def inference_campaign(weights, records, data_dir, output, names, deadline):
    output = Path(output)
    output.mkdir(parents=True, exist_ok=True)
    detector = YOLO(str(weights))
    # Single-image rectangular preprocessing is intentionally recorded separately from native batch8 val.
    detector.predict(np.zeros((1280, 1280, 3), np.uint8), imgsz=1280, device=0, conf=.001,
                     rect=True, quantize=32, max_det=300, verbose=False)
    cached = {mode: [] for mode in ('full', 'tile512', 'tile768')}
    try:
        for index, record in enumerate(records):
            if time.monotonic() >= deadline:
                raise TimeoutError('Campaign wall-clock budget reached during inference')
            image = cv2.imread(str(Path(data_dir) / record['image_path']))
            if image is None:
                raise ValueError('Unreadable validation image: ' + record['image_path'])
            height, width = image.shape[:2]
            classes, boxes = ground_truth(Path(data_dir) / record['label_path'], width, height)
            torch.cuda.synchronize()
            started = time.perf_counter()
            native = detector.predict(image, imgsz=1280, device=0, conf=.001, rect=True, quantize=32,
                                      max_det=300, verbose=False)[0].boxes.data.cpu().numpy()
            torch.cuda.synchronize()
            full_seconds = time.perf_counter() - started
            for mode, tile in [('full', None), ('tile512', 512), ('tile768', 768)]:
                predictions = [native]
                seconds = full_seconds
                if tile is not None:
                    torch.cuda.synchronize()
                    started = time.perf_counter()
                    for y in tile_starts(height, tile):
                        for x in tile_starts(width, tile):
                            if x == y == 0 and width <= tile and height <= tile:
                                continue  # identical full frame, no extra prediction
                            crop = image[y:min(y+tile, height), x:min(x+tile, width)]
                            found = detector.predict(crop, imgsz=1280, device=0, conf=.001, rect=True,
                                                     quantize=32, max_det=300, verbose=False)[0].boxes.data.cpu().numpy().copy()
                            found[:, [0, 2]] += x
                            found[:, [1, 3]] += y
                            predictions.append(found)
                    torch.cuda.synchronize()
                    seconds += time.perf_counter() - started
                folder = output / 'prediction_cache' / mode
                folder.mkdir(parents=True, exist_ok=True)
                path = folder / f'{index:05d}.npz'
                np.savez_compressed(path, pred=np.concatenate(predictions), classes=classes, boxes=boxes, seconds=seconds)
                cached[mode].append(path)
            if (index+1) % 100 == 0:
                print('INFERENCE_PROGRESS', index+1, '/', len(records), flush=True)
        result = {}
        for mode, paths in cached.items():
            for merge_iou in ([None, .6, .7, .8] if mode == 'full' else [.6, .7, .8]):
                if time.monotonic() >= deadline:
                    raise TimeoutError('Campaign wall-clock budget reached during metrics')
                label = mode + '_merge' + str(merge_iou)
                result[label] = summarize_cache(paths, names, merge_iou)
                (output / 'inference_comparison.json').write_text(json.dumps(result, indent=2), encoding='utf-8')
                print('INFERENCE_METRICS', label, {k:v for k,v in result[label].items() if isinstance(v,(float,int))}, flush=True)
        return result
    finally:
        del detector
        torch.cuda.empty_cache()


"""Six controlled trials from checkpoint 07, Project8 v0.4 best epoch 19."""


def build_matrix(initial, data, output, backbone_layers):
    if not backbone_layers or any(type(i) is not int or i < 0 for i in backbone_layers):
        raise ValueError('Supply verified backbone layer indices')
    common = dict(model=str(initial), data=str(data), project=str(output),
                  epochs=10, patience=3, resume=False, exist_ok=False,
                  seed=42, deterministic=True, pretrained=True, cls_remap=False,
                  imgsz=1280, batch=32, device='0,1', workers=2, cache=False,
                  amp=True, val=True, plots=True, save=True, save_period=1,
                  optimizer='MuSGD', lr0=.0001, lrf=.01, cos_lr=False,
                  warmup_epochs=1., warmup_bias_lr=.0001, warmup_momentum=.8,
                  momentum=.937, weight_decay=.0005, nbs=64,
                  hsv_h=.015, hsv_s=.7, hsv_v=.4, degrees=0., translate=.1,
                  scale=.5, shear=0., perspective=0., flipud=0., fliplr=.5,
                  mosaic=.5, close_mosaic=2, mixup=0., cutmix=0., copy_paste=0.,
                  box=7.5, cls=.5, dfl=1.5, cls_pw=0., multi_scale=0.,
                  rect=False, fraction=1., compile=False, freeze=None)
    changes = [dict(name='E1_MuSGD_lr1e4'), dict(name='E2_MuSGD_lr3e4', lr0=.0003),
               dict(name='E3_AdamW_lr1e4', optimizer='AdamW'),
               dict(name='E4_AdamW_lr3e4', optimizer='AdamW', lr0=.0003),
               dict(name='E5_MuSGD_scale02', scale=.2),
               dict(name='E6_MuSGD_frozen_backbone', freeze=list(backbone_layers))]
    return [dict(common, **change) for change in changes]

"""Sequential exploratory training; execute after verified staging cells."""
import inspect
import time
import traceback
import os
import signal
import subprocess
from ultralytics.utils.torch_utils import EarlyStopping

assert START_TRAINING and ALLOW_UNREVIEWED_PILOT
assert EPOCHS == 7 and PATIENCE == 0
assert torch.cuda.is_available()
assert torch.cuda.device_count() >= 2
assert file_sha(initial) == BEST_SHA256
assert ultralytics.__version__ == '8.4.165'

CHECKPOINT_NAME = 'checkpoint 07 — YOLO26n Project8 v0.4, phiên Kaggle ngày 08/10/2026, best epoch 19'
print(CHECKPOINT_NAME, initial, BEST_SHA256, flush=True)
print('EXPLORATORY: unreviewed v0.5 labels; no classroom improvement claim; no test selection.', flush=True)

probe = YOLO(str(initial))
assert probe.model.yaml.get('scale') == 'n'
backbone_count = len(probe.model.yaml['backbone'])
assert 0 < backbone_count < len(probe.model.model)
backbone_layers = list(range(backbone_count))
print('Verified backbone:', [(i, type(probe.model.model[i]).__name__) for i in backbone_layers], flush=True)
del probe; gc.collect(); torch.cuda.empty_cache()

SOURCE_HASHES={'E2_MuSGD_lr3e4':('7f214f51576abce18704a7434822d6db8bed5f14a50ff13e0a03eeb3623758cb','853093fa749a1c4dc600c4703929542b2ba2c127c7b3072b16df4896af2e9854'),'E3_AdamW_lr1e4':('9474d7f2b4c869e2101cf9f4f64b0f7f0411b859a8d31d3b2425aba48f479a13','763af06efb414ef0a032bb03db47b19a08287c4ae73a156f92f15ce4ac30b65e')}
MODEL_INPUT_ROOT = Path('/kaggle/input')  # Independent of extracted dataset root.
initials={}; jobs=[]
for source,index in [('E3_AdamW_lr1e4',2)]:
    weight_hash,archive_hash=SOURCE_HASHES[source]
    candidates=[p for p in MODEL_INPUT_ROOT.rglob('best.pt') if source in str(p) and file_sha(p)==weight_hash]
    if candidates: initials[source]=candidates[0]
    else:
        bundles=list(MODEL_INPUT_ROOT.rglob(source+'.zip')); assert len(bundles)==1,'Attach completed CP07 matrix notebook output: '+source
        assert file_sha(bundles[0])==archive_hash
        target=OUT/'initials'/source/'best.pt'; target.parent.mkdir(parents=True,exist_ok=True)
        with zipfile.ZipFile(bundles[0]) as archive: target.write_bytes(archive.read('weights/best.pt'))
        initials[source]=target
    assert file_sha(initials[source])==weight_hash
    detector=YOLO(str(initials[source])); assert detector.task=='detect' and list(detector.names.values())==NAMES
    assert all(torch.isfinite(v).all() for v in detector.model.state_dict().values() if v.is_floating_point())
    del detector; gc.collect(); torch.cuda.empty_cache()
    config=build_matrix(initials[source],DATA_YAML,OUT/'train',backbone_layers)[index]
    base = dict(config, epochs=7, patience=0, imgsz=1280, batch=32, save_period=-1, bgr=0.)
    groups = [
        ('CONTROL', {}, []),
        ('LIGHT_CONTRAST', {}, [A.RandomBrightnessContrast(brightness_limit=.15, contrast_limit=.20, p=.30),
                               A.RandomGamma(gamma_limit=(85,115), p=.15), A.CLAHE(clip_limit=(1,2), p=.05)]),
        ('GEOMETRY', dict(degrees=8., shear=1., perspective=.0001), []),
        ('GRAY', {}, [A.ToGray(p=.15)]),
        ('INVERT', {}, [A.InvertImg(p=.05)]),
    ]
    for group, changes, transforms in groups:
        serialized = [A.to_dict(t) for t in transforms]
        from ultralytics.data.augment import Albumentations as YOLOAlbumentations
        aug_probe = YOLOAlbumentations(transforms=serialized)
        assert aug_probe.transform is not None and not aug_probe.contains_spatial
        assert len(aug_probe.transform.transforms) == len(transforms)
        jobs.append(dict(base, **changes, name=source+'_'+group+'_1280_E7_B32', augmentations=serialized))
    print('INITIAL_VERIFIED', source, weight_hash, flush=True)

assert len(jobs)==5
assert all(j['epochs']==7 and j['imgsz']==1280 and j['batch']==32 and j['device']=='0,1' for j in jobs)
assert all(j['model']==str(initials['E3_AdamW_lr1e4']) and j['optimizer']=='AdamW' and j['lr0']==.0001 for j in jobs)
print('CAMPAIGN_PREFLIGHT_PASS original E3; five independent runs,7epoch each,1280,batch32,T4x2', flush=True)
print('CAMPAIGN_LIST', json.dumps(jobs, indent=2), flush=True)

started = time.monotonic()
state = dict(status='preflight_pass', checkpoint_name=CHECKPOINT_NAME,
             initial_sha256=BEST_SHA256, manifest_sha256=MANIFEST_SHA256,
             full_visual_review_complete=bool(TRAIN_GATE),
             user_authorized_unreviewed_method_experiment=True,
             classroom_claim_allowed=False, test_used=False, jobs=jobs, runs={},
             environment=dict(ultralytics=ultralytics.__version__, torch=torch.__version__,
                              cuda=torch.version.cuda, gpus=[torch.cuda.get_device_name(i) for i in (0, 1)]),
             source_hashes=SOURCE_HASHES,
             evaluation_note='Primary: native before/after1280 with nmsFalse, batch8, quantize32, rectTrue; trainer final is reported separately. No held-out test used.')


def preserve_state():
    state['elapsed_seconds'] = round(time.monotonic()-started, 1)
    temporary = OUT/'status.json.partial'
    temporary.write_text(json.dumps(state, indent=2, ensure_ascii=False, allow_nan=False), encoding='utf-8')
    temporary.replace(OUT/'status.json')


def archive_trial(label):
    trial = OUT/'train'/label
    if not trial.exists():
        return
    bundle = Path(shutil.make_archive(str(OUT/label), 'zip', root_dir=trial))
    with zipfile.ZipFile(bundle) as archive:
        assert archive.testzip() is None
    state['runs'][label]['archive'] = dict(path=str(bundle), sha256=file_sha(bundle), bytes=bundle.stat().st_size)
    preserve_state()
    print('ARCHIVE_VERIFIED', label, file_sha(bundle), flush=True)


def native_eval(weights, label, size=1280):
    detector = YOLO(str(weights))
    metrics = detector.val(data=str(DATA_YAML), split='val', imgsz=size,
                           batch=8, device=0, workers=2, conf=.001, iou=.7,
                           nms=False, max_det=300, augment=False, quantize=32, rect=True, plots=False,
                           project=str(OUT/'evaluation'), name=label, exist_ok=False)
    result = dict(precision=float(metrics.box.mp), recall=float(metrics.box.mr), f1_macro_at_validator_operating_point=float(metrics.box.f1.mean()), harmonic_of_mean_PR=float(2*metrics.box.mp*metrics.box.mr/(metrics.box.mp+metrics.box.mr)) if metrics.box.mp+metrics.box.mr else 0., speed=metrics.speed, mAP50=float(metrics.box.map50), mAP50_95=float(metrics.box.map),
                  per_class=json.loads(json.dumps(metrics.summary(), default=lambda v: v.item())), per_class_AP50_95=dict(zip(NAMES, map(float, metrics.box.maps))),
                  protocol=dict(imgsz=size, quantize=32, rect=True, batch=8, device=0, max_det=300, augment=False, conf=.001, iou=.7, nms=False, split='val'),
                  checkpoint_sha256=file_sha(weights))
    del detector; gc.collect(); torch.cuda.empty_cache()
    return result


def train_ddp(config, record):
    """CLI spawns DDP; parent reads persisted epoch history, not child callbacks."""
    config_path = OUT/(config['name']+'_config.yaml')
    config_path.write_text(yaml.safe_dump(config, sort_keys=False), encoding='utf-8')
    log_path = OUT/(config['name']+'_training.log')
    history_path = Path(config['project'])/config['name']/'results.csv'
    launched = time.monotonic()
    record['log'] = str(log_path)
    with log_path.open('w', encoding='utf-8') as log:
        process = subprocess.Popen([shutil.which('yolo'), 'detect', 'train', 'cfg='+str(config_path)],
                                   stdout=log, stderr=subprocess.STDOUT, start_new_session=True)
        previous_epoch = 0
        last_heartbeat = time.monotonic()
        print('TRAIN_LAUNCHED',config['name'],'PID',process.pid,'epochs',config['epochs'],'imgsz',config['imgsz'],'batch',config['batch'],'device',config['device'],flush=True)
        while process.poll() is None:
            if time.monotonic()-last_heartbeat >= 60:
                log_bytes = log_path.stat().st_size
                with log_path.open('rb') as live_log:
                    live_log.seek(max(0,log_bytes-2000))
                    log_lines = live_log.read().decode('utf-8',errors='replace').splitlines()
                latest_line = next((line.strip() for line in reversed(log_lines) if line.strip()),'')
                print('TRAIN_HEARTBEAT',config['name'],'seconds',round(time.monotonic()-launched,1),'completed_epoch',previous_epoch,'log_bytes',log_bytes,'latest',latest_line[-500:],flush=True)
                last_heartbeat=time.monotonic()

            if history_path.is_file():
                with history_path.open(newline='') as stream:
                    history = [{k.strip(): v.strip() for k, v in row.items()} for row in csv.DictReader(stream)]
                # Ignore a last row still being written; trainer writes metrics before weights.
                if history and history[-1].get('metrics/mAP50-95(B)'):
                    latest = history[-1]
                    epoch = int(latest['epoch'])
                    fitness = float(latest['metrics/mAP50-95(B)'])
                    if not math.isfinite(fitness) or epoch > config['epochs']:
                        os.killpg(process.pid, signal.SIGTERM)
                        process.wait()
                        raise ValueError('Invalid fitness or epoch cap exceeded')
                    if epoch != previous_epoch:
                        record.update(status='running', epochs_completed=epoch, fitness=fitness,
                                      epoch_metrics=history, seconds=time.monotonic()-launched)
                        previous_epoch = epoch
                        preserve_state()
                        print('PROGRESS_FULL', config['name'], json.dumps(latest), flush=True)
            if time.monotonic()-launched >= 60*60 or time.monotonic()-started >= CAMPAIGN_BUDGET_HOURS*3600:
                record['budget_stop'] = True
                os.killpg(process.pid, signal.SIGTERM)
                try:
                    process.wait(timeout=30)
                except subprocess.TimeoutExpired:
                    os.killpg(process.pid, signal.SIGKILL)
                    process.wait()
                record['status'] = 'budget_stop'
                preserve_state()
                return
            time.sleep(10)
        if process.returncode != 0:
            print(log_path.read_text(encoding='utf-8')[-8000:], flush=True)
            raise RuntimeError('DDP training failed; exit '+str(process.returncode))
    with history_path.open(newline='') as stream:
        history = [{k.strip(): v.strip() for k, v in row.items()} for row in csv.DictReader(stream)]
    record.update(epochs_completed=len(history), epoch_metrics=history, seconds=time.monotonic()-launched)
    assert 0 < len(history) <= config['epochs']
    print(log_path.read_text(encoding='utf-8')[-3500:], flush=True)


preserve_state()
state['baseline_native1280']=native_eval(initials['E3_AdamW_lr1e4'], 'original_native1280')
preserve_state()
state['status']='confidence_and_tiles'
preserve_state()
try:
    state['inference']=inference_campaign(initials['E3_AdamW_lr1e4'], [r for r in train_val if r['split']=='val'],
                                        DATA_DIR, OUT/'inference', NAMES, started+CAMPAIGN_BUDGET_HOURS*3600)
except Exception:
    state.update(status='inference_failed_requires_review', error=traceback.format_exc())
    preserve_state()
    raise
preserve_state()
for config in jobs:
    label=config['name']; source=next(k for k in SOURCE_HASHES if label.startswith(k))
    initial=initials[source]; expected_initial_hash=SOURCE_HASHES[source][0]
    assert file_sha(initial)==expected_initial_hash
    assert time.monotonic()-started<CAMPAIGN_BUDGET_HOURS*3600
    state['current_run']=label; state['status']='before_validation'
    record=state['runs'][label]=dict(status='before_validation',initial_sha256=expected_initial_hash,epochs_completed=0,before={})
    preserve_state()
    for size in (1280,):
        record['before'][str(size)]=state['baseline_native1280']; preserve_state()
    state['status']='training'; record['status']='starting'; preserve_state()
    try:
        train_ddp(config,record)
        assert not record.get('budget_stop') and record['epochs_completed']==7
        best=Path(config['project'])/label/'weights/best.pt'; last=best.with_name('last.pt')
        assert best.is_file() and last.is_file()
        record.update(status='training_completed',best_sha256=file_sha(best),last_sha256=file_sha(last))
        archive_trial(label)
        record['after_best']={}
        for size in (1280,):
            record['after_best'][str(size)]=native_eval(best,label+'_after_best_'+str(size),size); preserve_state()
        record['after_last_1280']=native_eval(last,label+'_after_last_1280',1280)
        record['status']='completed'; preserve_state()
        print('NATIVE_BEST_FULL', label, json.dumps(record['after_best']['1280']), flush=True)
    except Exception:
        record.update(status='failed',error=traceback.format_exc()); state['status']='failed_requires_review'; preserve_state(); archive_trial(label); raise
state['status']='completed'; preserve_state()
print('E3_ORIGINAL_CAMPAIGN_COMPLETE confidence/tile comparison and five independent7epoch trials. Unreviewed methodology only.',flush=True)
```
